## Data Quality Limitation

During the data cleaning process, it was identified that some records contain shipping dates and delivery dates earlier than the order date, which is logically inconsistent. Therefore, shipping time and delivery time analyses were not performed to avoid producing misleading results.

```python
# Finding outliers using Box plot chart
import matplotlib.pyplot as plt
plt.figure(figsize=(8,3))
plt.boxplot(df["total_sales"])
plt.title("BoxPlot of Total Sales")
plt.show()
```

```python
#finding outliers Using IQR method
Q1 = df["total_sales"].quantile(0.25)
Q3 = df["total_sales"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print(lower_bound)
print(upper_bound)
```

```python
# How many outliers are there 
(df["total_sales"] > upper_bound).sum()
```

# Time Analysis

#### Dataset Limitation
**Note:** The dataset contains transaction records from 1 January 2026 to 10 February 2026 only. Therefore, year-wise and quarter-wise sales analysis cannot be performed because the data covers only one year and one quarter.

### Q1: Which month generated the highest total sales?

```python
monthly_sales = df.groupby(df["order_date"].dt.month_name())["total_sales"].sum().sort_values(ascending=False)
monthly_sales
```

#### Insight: 
Insight: January generated the highest total sales (₹489,990,100), while February generated ₹152,139,000. January was the best-performing month in terms of revenue during the analysis period.

### Q2: Which day of the week generated the highest total sales?

```python
days_sales = df.groupby(df["order_date"].dt.day_name())["total_sales"].sum().sort_values(ascending=False)
days_sales
```

#### Insight:
Monday generated the highest total sales (₹95,371,920), while Wednesday recorded the lowest sales among all days of the week. This indicates that customer purchasing activity was strongest at the beginning of the week during the analysis period.

### Q3: Which month received the highest number of orders?

```python
order_by_month = df.groupby(df["order_date"].dt.month_name())["order_id"].count().sort_values(ascending=False)
order_by_month
```

#### Insight:
January recorded the highest number of orders (7653), followed by February (2347).
